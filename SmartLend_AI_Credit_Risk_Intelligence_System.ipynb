```python
# =========================================================

# PART 1 — DATA CLEANING & FEATURE ENGINEERING
# =========================================================

# =========================================================
# IMPORT LIBRARIES
# =========================================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler

import warnings
warnings.filterwarnings('ignore')

# =========================================================
# PROFESSIONAL NOTEBOOK STYLE
# =========================================================

plt.style.use('dark_background')

sns.set_style("whitegrid")

plt.rcParams['figure.figsize'] = (10,6)

plt.rcParams['font.size'] = 12

# =========================================================
# LOAD DATASET
# =========================================================

# Updated filename to match available resource in /content/
df = pd.read_excel('/content/SmartLend AI.csv.xlsx')

print("="*60)
print("SMARTLEND AI CREDIT RISK INTELLIGENCE SYSTEM")
print("="*60)

# =========================================================
# DATA INSPECTION
# =========================================================

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())

# =========================================================
# MISSING VALUE ANALYSIS
# =========================================================

print("\nMissing Values:\n")

display(df.isnull().sum())

plt.figure(figsize=(14,6))

sns.heatmap(
    df.isnull(),
    cmap='viridis',
    cbar=False
)

plt.title(
    'Missing Values Heatmap',
    fontsize=16,
    fontweight='bold'
)

plt.show()

# =========================================================
# REMOVE DUPLICATES
# =========================================================

print("\nDuplicate Rows:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Duplicates Removed Successfully")

# =========================================================
# HANDLE MISSING VALUES
# =========================================================

numerical_columns = df.select_dtypes(
    include=np.number
).columns

categorical_columns = df.select_dtypes(
    exclude=np.number
).columns

# Numerical Columns
for col in numerical_columns:

    df[col].fillna(
        df[col].median(),
        inplace=True
    )

# Categorical Columns
for col in categorical_columns:

    df[col].fillna(
        df[col].mode()[0],
        inplace=True
    )

print("\nMissing Values Handled Successfully")

# =========================================================
# OUTLIER DETECTION
# =========================================================

for col in numerical_columns:

    plt.figure(figsize=(8,4))

    sns.boxplot(
        x=df[col],
        color='cyan'
    )

    plt.title(
        f'Outlier Detection — {col}',
        fontsize=14,
        fontweight='bold'
    )

    plt.show()

# =========================================================
# IQR OUTLIER TREATMENT
# =========================================================

for col in numerical_columns:

    Q1 = df[col].quantile(0.25)

    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)

    upper = Q3 + (1.5 * IQR)

    df[col] = np.where(
        df[col] < lower,
        lower,
        df[col]
    )

    df[col] = np.where(
        df[col] > upper,
        upper,
        df[col]
    )

print("\nOutlier Treatment Completed")

# =========================================================
# FEATURE ENGINEERING
# =========================================================

df['Loan_to_Income_Ratio'] = (
    df['LoanAmount'] / df['AnnualIncome']
)

df['EMI_Burden_Percent'] = (
    (df['MonthlyEMI'] / df['MonthlyIncome']) * 100
)

df['Credit_Utilization_Flag'] = np.where(
    df['CreditUtilization'] > 60,
    1,
    0
)

print("\nFeature Engineering Completed")

# =========================================================
# LABEL ENCODING
# =========================================================

encoder = LabelEncoder()

for col in categorical_columns:

    df[col] = encoder.fit_transform(df[col])

print("\nCategorical Encoding Completed")

# =========================================================
# FEATURE SCALING
# =========================================================

target_col = 'Default'

numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

if target_col in numerical_columns:

    numerical_columns.remove(target_col)

scaler = StandardScaler()

df[numerical_columns] = scaler.fit_transform(
    df[numerical_columns]
)

print("\nFeature Scaling Completed")

# =========================================================
# SAVE CLEANED DATASET
# =========================================================

df.to_csv(
    'cleaned_finloan_data.csv',
    index=False
)

print("\nCleaned Dataset Saved Successfully")
```
